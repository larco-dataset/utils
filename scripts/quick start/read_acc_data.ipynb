```python
sample_lab = lab_data[lab_data["type"] == "wm"].iloc[0]["file_path"] # wm = washing machine
df_lab = pd.read_parquet(sample_lab)

print(df_lab.columns)

df_lab.plot(x="timestamp", y="back.x", figsize=(10,4), title="Lab Washing Machine Back X-axis Accelerometer Data")
plt.show()
```

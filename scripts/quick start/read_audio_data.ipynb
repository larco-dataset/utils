```python
import matplotlib.pyplot as plt
from scipy.io import wavfile

sample_lab = lab_data[lab_data["type"] == "wm"].iloc[0]["file_path"] # wm = washing machine

print(sample_lab)

# Load wav file
sample_rate, data = wavfile.read(sample_lab)

# If stereo, take one channel
if data.ndim > 1:
    data = data[:, 0]

# Create time axis in seconds
time = [i / sample_rate for i in range(len(data))]

# Plot
plt.figure(figsize=(12, 4))
plt.plot(time, data)
plt.title("Waveform")
plt.xlabel("Time [s]")
plt.ylabel("Amplitude")
plt.show()
```
